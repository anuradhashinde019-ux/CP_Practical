## 2. What is a Factorial?

For a non-negative integer `n`, factorial is written as `n!`.

For example:

```text
5! = 5 × 4 × 3 × 2 × 1 = 120
```

The recursive definition is:

```text
0! = 1
n! = n × (n-1)!    for n > 0
```

## 4. Activity 2 – Recursive Factorial Function

Create a Python function:

```text

```

The function should:

1. Check the base case.
2. Return `1` for `n = 0`.
3. Otherwise return `n * factorial(n-1)`.

Test it with `0`, `1`, `3`, `5` and `7`.

### Copilot Activity 2 – Write the Recursive Logic

Ask Copilot:

> I need to implement factorial recursively in Python. Explain how the function call factorial(n-1) eventually reaches the base case. Do not give me the complete code.

````python
# Activity 2
# Write your recursive factorial function here.
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
````

## 7. Activity 5 – Factorial and Combinations

Factorials are useful in combinatorics.

The number of ways of selecting `r` objects from `n` objects, when order does not matter, is:

```text
C(n, r) = n! / (r! (n-r)!)
```

These combinations are useful when discussing possible selections of molecules, components, samples or experimental choices.

For example, if 5 different chemical components are available and we select 2 of them, the number of possible selections is:

```text
C(5, 2)
```

### Copilot Activity 5 – Understand Combinations

Ask Copilot:

> Explain the combination formula C(n,r) = n!/(r!(n-r)!) to a second-year chemical engineering student. Explain why factorial appears in the formula. Do not write Python code.

````python
# Activity 5
# Use your recursive factorial function to calculate C(5, 2).
def combination(n,r):
    if r > n:
        return 0
    return factorial(n) // (factorial(r) * factorial(n - r))
````

````python
combination_result = combination(5,2)
print("c(5,2) =",combination_result)
````

Use:


### Engineering interpretation

The result tells us how many different **sets of 3 components** can be selected from the 6 candidates, assuming that:

- each component can be selected at most once,
- order does not matter,
- every combination is considered separately.

This is a simplified combinatorial example, not a statement about actual formulation feasibility.
