## 7. Activity 5 – Filter Experimental Records

Use Pandas filtering to answer questions such as:

### Question A
Show only observations where:

```text
Temperature_C > 30
```

### Question B
Show only records from `EXP001`.

### Question C
Show observations where temperature is above 30 °C **and** pressure is greater than 1.05 bar.

### Think

Why is filtering important when analyzing experimental data?

````python
# Activity 5
# Write Pandas filtering expressions for Questions A, B and C.
# Question A Show only observations where:```textTemperature_C > 30
filtered_df_A = df_new[df_new['Temperature_C'] > 30]
print("\nObservations where Temperature_C > 30:")
print(filtered_df_A)
print('_ _ _ _ _')

# Question B Show only records from `EXP001`.
print("\nObservations from EXP001:")
filtered_df_B = df_new[df_new['Experiment_ID'] == 'EXP001']
print(filtered_df_B)
print('_ _ _ _ _')

# Question C Show observations where temperature is above 30 °C **and** pressure is greater than 1.05 bar.
print("\nObservations where Temperature_C > 30 and Pressure_bar > 1.05:")
filtered_df_C = df_new[(df_new['Temperature_C'] > 30) & (df_new['Pressure_bar'] > 1.05)]
print(filtered_df_C)    
print('_ _ _ _ _')

````

## 12. Activity 10 – Simple Experimental Data Summary

Use Pandas to calculate:

1. Average temperature for each experiment.
2. Maximum pressure for each experiment.
3. Final concentration for each experiment.

Use grouping by `Experiment_ID`.

### Engineering interpretation

This is a basic example of converting raw experimental logs into a compact summary that an engineer could inspect before further analysis.

````python
# Activity 10
# Create an experiment-level summary using groupby().
df_summary = df_new.groupby('Experiment_ID').agg({
    'Temperature_C': ['mean'],
    'Pressure_bar': ['max'],
    'Concentration_mol_L': ['last']})

print("\nExperiment-level summary:")
print(df_summary)
````
