## 6. Activity 4 – Floating-Point Accuracy

Python uses floating-point representation for many decimal values. Because of the way computers store binary floating-point numbers, some apparently simple decimal calculations can produce a very small representation difference.

Run the following expressions:

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

Observe the result.

This does **not** mean Python cannot perform decimal calculations. It means that binary floating-point representation can introduce very small rounding differences.

## 7. Activity 5 – Rounding Chemical Measurements

Laboratory results are often reported with an appropriate number of decimal places.

Use Python's `round()` function to display a calculated concentration to:

- 2 decimal places
- 3 decimal places
- 4 decimal places

For example, compare the raw calculation with its rounded display.

**Important:** Rounding the displayed value does not necessarily mean that the original stored value has been changed.


````python
# Activity 5
# Calculate a chemical measurement and display it with different precision.
measurement = 10.12345678  # Example chemical measurement value
print(f"Measurement:{measurement}") 
print(f"Measurement(2 decimal places):{measurement:.2f}")
print(f"Measurement(4 decimal places):{measurement:.4f}")
print(f"Measurement(6 decimal places):{measurement:.6f}")

m2=round(measurement,2)
print(f"Measurement(2 decimal places using round()):{m2}")
m3=round(measurement,3)
print(f"Measurement(3 decimal places using round()):{m3}")
m4=round(measurement,4)
print(f"Measurement(4 decimal places using round()):{m4}")
````

````python
# FINAL PROGRAM
# Chemical Laboratory Arithmetic & Measurement Calculator
print("CHEMICAL LABORATORY CALCULATOR")
print("1. Add two measurements")
print("2. Subtract two measurements")
print("3. Multiply two measurements")
print("4. Divide two measurements")
print("5. Find remainder")
print("6. Calculate concentration (mass/volume)")

choice = input("Enter your choice (1-6): ")

if choice not in ("1", "2", "3", "4", "5", "6"): 
   print("Error: Please choose a number from 1 to 6.")
else:  
   try:       
       if choice in ("1", "2", "3", "4", "5"): 
           value_1 = float(input("Enter the first measurement: "))     
           value_2 = float(input("Enter the second measurement: "))  
           print(f"You entered: {value_1} and {value_2}")    
       else:          
           mass = float(input("Enter the mass (g): "))     
           volume = float(input("Enter the volume (mL): "))          
           print(f"You entered mass: {mass} g and volume: {volume} mL")   
   except ValueError:      
                print("Error: Please enter valid numerical values only.")

   else:    
        if choice == "1":        
           result = value_1 + value_2          
           operation = "Addition"        
           print(f"{operation} result: {result:.4f}")   

        elif choice == "2":           
           result = value_1 - value_2          
           operation = "Subtraction"          
           print(f"{operation} result: {result:.4f}")     
           
        elif choice == "3":       
             result = value_1 * value_2        
             operation = "Multiplication"         
             print(f"{operation} result: {result:.4f}")

        elif choice == "4":         
             if value_2 == 0:             
                   print("Error: Division by zero is not allowed.")     
             else:           
                  result = value_1 / value_2        
                  operation = "Division"              
                  print(f"{operation} result: {result:.4f}")   
                  
        elif choice == "5":           
             if value_2 == 0:         
                print("Error: Modulus by zero is not allowed.")          
             else:              
                  result = value_1 % value_2         
                  operation = "Modulus"             
                  print(f"{operation} result: {result:.4f}")      
                  
        elif choice == "6":           
             if volume == 0:              
                  print("Error: Volume cannot be zero.")       
             else:               
                  result = mass / volume
                  operation = "Concentration"
                  print(f"{operation} result: {result:.4f} g/mL")
                  
````
